## Echo mini cram sheet

### 1) Inheritance

A child class inherits attributes and methods from a parent class.

```python
class Parent:
    def say_hi(self):
        print("Hola!")

class Child(Parent):
    pass
```

Memory hook: **child gets parent stuff**.

### 2) `isinstance()`

Checks whether an object belongs to a class, including inherited relationships.

```python
class Parent:
    pass

class Child(Parent):
    pass

c = Child()
print(isinstance(c, Child))   # True
print(isinstance(c, Parent))  # True
```

### 3) `issubclass()`

Checks whether one class inherits from another.

```python
print(issubclass(Child, Parent))  # True
```

### 4) Overriding

A child can replace a parent method.

```python
class Parent:
    def say_hi(self):
        print("Hola!")

class Child(Parent):
    def say_hi(self):
        print("Hi!")
```

Memory hook: **override = replace the parent version**.

### 5) `super()`

Use `super()` to call the next class method in the lookup chain.

```python
class Parent:
    def say_hi(self):
        return "Hola!"

class Child(Parent):
    def say_hi(self):
        return super().say_hi() + " Hi!"
```

Memory hook: **`super()` = go to the next class in the MRO**.

### 6) Constructor chaining

When the child has its own `__init__`, use `super().__init__(...)` to initialize the parent part.

```python
class Parent:
    def __init__(self, name):
        self.name = name

class Child(Parent):
    def __init__(self, name, language):
        super().__init__(name)
        self.language = language
```

Rule: pass only what the parent constructor expects.

### 7) Python does **not** support normal method overloading

In Python, redefining a method with the same name replaces the old one.

```python
class Calc:
    def add(self, *nums):
        return sum(nums)
```

### 8) MRO = Method Resolution Order

MRO is the order Python uses to search for attributes and methods.

```python
class Mother:
    pass

class Father:
    pass

class Child(Mother, Father):
    pass

print(Child.mro())
```

This gives the search order:

**Child -> Mother -> Father -> object**

### 9) Big exam warning

**`super()` follows the MRO, not just the direct parent.**

Important example:

```python
class Parent:
    def display(self): print("PARENT")

class A(Parent):
    def display(self):
        super().display()
        print("A")

class B(Parent):
    def display(self):
        super().display()
        print("B")

class C(A, B):
    def display(self):
        super().display()
        print("C")

C().display()
```

Output:

```python
PARENT
B
A
C
```

Why?

MRO is:

**C -> A -> B -> Parent -> object**

So when `A` calls `super()`, it goes to **B**, not directly to `Parent`.

### 10) Common MRO pattern

```python
class A: pass
class B(A): pass
class C(A): pass
class D(B, C): pass

print(D.mro())
```

Result:

**D -> B -> C -> A -> object**

### 11) Error case to remember

This can fail:

```python
class F: pass
class E(F): pass
class G(F, E): pass
```

Reason: Python cannot create a consistent MRO.

### 12) One caution

This example is risky / likely wrong if written carelessly:

```python
class Mother:
    def __init__(self):
        self.a = 1

class Father:
    def __init__(self):
        self.b = 2

class Child(Mother, Father):
    pass
```

`Child()` will not automatically run both parent constructors unless the classes are designed to cooperate with `super()` correctly.

### 13) 5-second memory list

- **Inheritance** = child gets parent stuff
- **Override** = child replaces parent method
- **`isinstance()`** = object check
- **`issubclass()`** = class check
- **`super()`** = next class in MRO
- **MRO** = Python search order
- **Multiple inheritance** = powerful but dangerous

- trace `super()`
