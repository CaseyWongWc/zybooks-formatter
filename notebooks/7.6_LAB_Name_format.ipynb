## 🔑 The key concept: `.split()`

`"Pat Silly Doe".split()` → gives you a **list**: `["Pat", "Silly", "Doe"]`

Then you can grab pieces:
- `parts[0]` → first name → `parts[0][0]` → first initial `"P"`
- `parts[-1]` → last name (always works whether 2 or 3 words!)
- `parts [drive.google](https://drive.google.com/file/d/14vpQsruhJBYDfBiXt8KjW1twB-HwsGJN/view?usp=drivesdk)` → middle name (only exists if 3 words)

## 🧠 Your logic plan

```python
full_name = input()
parts = full_name.split()

if len(parts) == 3:
    # firstName middleName lastName
    print(f"{parts[-1]}, {parts[0][0]}.{parts [drive.google](https://drive.google.com/file/d/14vpQsruhJBYDfBiXt8KjW1twB-HwsGJN/view?usp=drivesdk)}.")
else:
    # firstName lastName
    print(f"{parts[-1]}, {parts[0][0]}.")
```
