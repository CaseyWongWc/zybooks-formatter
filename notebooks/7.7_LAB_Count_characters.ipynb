## 7.7 LAB: Count characters

### LAB ACTIVITY: LAB: Count characters

Write a program whose input is a string which contains a character and a phrase, and whose output indicates the number of times the character appears in the phrase. The output should include the input character and use the plural form, n's, if the number of times the characters appears is not exactly 1.
Ex: If the input is:

```python
n Monday
```

the output is:

```python
1 n
```

Ex: If the input is:

```python
z Today is Monday
```

the output is:

```python
0 z's
```

Ex: If the input is:

```python
n It's a sunny day
```

the output is:

```python
2 n's
```

Case matters. n is different than N.
Ex: If the input is:

```python
n Nobody
```

the output is:

```python
0 n's
```

**Test Cases:**
| # | Input | Expected Output | Points |
|---|-------|-----------------|--------|
| 1 | `n Monday` | `1 n` | 2 |
| 2 | `z Today is Monday` | `0 z's` | 2 |
| 3 | `n It's a sunny day` | `2 n's` | 2 |
| 4 | `t this is a sentence with many t's` | `4 t's` | 2 |
| 5 | `x X` | `0 x's` | 2 |
*Total: 10 points*
