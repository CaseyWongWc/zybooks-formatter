### Problem 8: Sum of evens

Write a function `sum_evens(n)` that returns the sum of all even numbers from 2 to n (inclusive).

Example: `sum_evens(6)` returns `12` (2+4+6)

```python
# ANSWER - don't peek until you've written it!
def sum_evens(n):
    total = 0
    for i in range(2, n + 1, 2):
        total += i
    return total

# Test
print(sum_evens(6))   # 12
print(sum_evens(10))  # 30
print(sum_evens(1))   # 0
```

### Problem 9: Count vowels

Write a function `count_vowels(s)` that returns the number of vowels (a,e,i,o,u) in a string. Case insensitive.

```python
# ANSWER - don't peek until you've written it!
def count_vowels(s):
    count = 0
    for ch in s.lower():
        if ch in 'aeiou':
            count += 1
    return count

# Test
print(count_vowels('Hello World'))  # 3
print(count_vowels('AEIOU'))        # 5
print(count_vowels('rhythm'))       # 0
```

### Problem 10: Reverse a string

Write a function `reverse_str(s)` that returns the reversed string WITHOUT using slicing or reversed().

```python
# ANSWER - don't peek until you've written it!
def reverse_str(s):
    result = ''
    for ch in s:
        result = ch + result
    return result

# Test
print(reverse_str('hello'))   # olleh
print(reverse_str('Python'))  # nohtyP
print(reverse_str(''))        # (empty)
```

### Problem 15: Write a function with dicts

Write a function `word_count(sentence)` that takes a string and returns a dict with each word as key and its count as value.

Example: `word_count('the cat and the dog')` returns `{'the': 2, 'cat': 1, 'and': 1, 'dog': 1}`

```python
# ANSWER - don't peek until you've written it!
def word_count(sentence):
    counts = {}
    for word in sentence.split():
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1
    return counts

# Test
print(word_count('the cat and the dog'))
print(word_count('hello hello hello'))
print(word_count('one'))
```
