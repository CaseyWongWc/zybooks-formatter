# Part A — Properties and Encapsulation

## Encapsulation

Encapsulation means outside code should not directly touch internal data. Instead, access is controlled through methods or properties.

Common patterns:

- `_name` means “internal use” by convention
- `__name` uses **name mangling**, which hides it more strongly

```python
class Person:
    def __init__(self, name="Unknown", age=0):
        self._name = name
        self._age = age
```

## Old way: getter and setter methods

```python
class Person:
    def __init__(self, name="Unknown", age=1):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def set_name(self, new_name):
        self.__name = new_name

p1 = Person("Mickey", 100)
print(p1.get_name())
p1.set_name("Mouse")
print(p1.get_name())
```

What to know:

- `get_name()` reads the value
- `set_name(...)` changes the value
- This is valid Python, but it is not the most Pythonic style

## New Pythonic way: `@property`

A property lets a method act like an attribute.

Instead of:

```python
print(p1.get_name())
```

you can do:

```python
#print(p1.name)
```

Example:

```python
class Person:
    def __init__(self, name="Unknown", age=1):
        self.__name = name
        self.__age = age

    @property
    def name(self):
        return self.__name
```

## Setter with `@name.setter`

```python
class Person:
    def __init__(self, name="Unknown", age=1):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        self._name = new_name
```

Usage:

```python
p1 = Person("Mickey")
print(p1.name)
p1.name = "Mouse"
print(p1.name)
```

This replaces:

```python
#p1.set_name("Mouse")
```

with:

```python
p1.name = "Mouse"
```

## Old way vs new way

Old way:

```python
#print(p1.get_name())
#p1.set_name("Mouse")
```

New way:

```python
print(p1.name)
p1.name = "Mouse"
```

## Read-only vs writable

A class can provide:

- a getter only → read-only
- a getter and setter → readable and writable

Example read-only:

```python
class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name
```

# Part B — Modules

## What is a module?

A **module** is a Python `.py` file that can contain:

- functions
- classes
- variables

Modules help split code into separate files instead of putting everything into one giant script.

## Types of modules

### Built-in modules

Examples:

- `math`
- `random`
- `os`
- `datetime`

### Third-party modules

Examples:

- `numpy`
- `pandas`
- `requests`

These usually need installation, often with `pip`.

### User-defined modules

Examples:

- `my_utils.py`
- `calculator.py`

## Basic import

```python
import math
import random
import os

print(math.sqrt(16))
print(math.pi)
print(random.randint(1, 10))
```

If you import the whole module, use `module_name.function_name`.

## Import with alias

```python
import math as m
import pandas as pd
import numpy as np

print(m.sqrt(25))
print(pd.__version__)
```

## Import specific items

```python
from math import sqrt, pi, pow

print(sqrt(36))
print(pi)
print(pow(2, 3))
```

## Import a specific item with alias

```python
from datetime import datetime as dt

today = dt.now()
print(today)
```

# Part C — Module behavior and `__name__`

## Importing a module runs its top-level code

If `module_a.py` contains:

```python
name = "Cal Poly Pomona"
print(f"From Module A: {name}")
```

then importing it can cause that print statement to run.

## `__name__`

- when a file is run directly, `__name__ == "__main__"`
- when a file is imported, `__name__` becomes the module name

Example:

```python
print(__name__)
print(type(__name__))
print(__name__ == "__main__")
```

Typical direct-run output:

```python
'''
__main__
<class 'str'>
True
'''
```

## Preventing import-side effects

Use:

```python
if __name__ == "__main__":
    print("run only when file is executed directly")
```

# Part F — Common mistakes and pitfalls

## Pitfall checklist

### Mistake 1: thinking `_name` is truly private

It is only a convention.

### Mistake 2: mixing `_name` and `__name`

These are not the same.

### Mistake 3: forgetting the setter decorator format

Correct:

```python
@name.setter
def name(self, new_name):
    self._name = new_name
```

### Mistake 4: forgetting module prefix after `import module`

If you do:

```python
import math
```

you must use:

```python
math.sqrt(16)
```

not just `sqrt(16)`.

### Mistake 5: forgetting import-side effects

If a module has top-level print statements, importing it can run them.

### Mistake 6: forgetting `if __name__ == "__main__":`

Without it, test/demo code may run during import.

### Mistake 7: using `from module import *`

Works, but is bad style.

### Mistake 8: confusing module names and variable names

`module_a` is the module.

`name` or `school` might be variables inside it.
