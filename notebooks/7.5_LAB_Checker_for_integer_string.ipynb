### 🧪 Now: 7.5 LAB — Checker for integer string

Here's what this lab asks you to do: **given a string, check if it represents a valid integer** (could be negative, so starts with `-` or just digits).

The key tools you'll use:
- `.isdigit()` — checks if all chars are digits
- string slicing `s[1:]` — to skip the `-` sign if present
- an `if/else` to handle both cases

Here's the starter logic to think about:

```python
user_string = input()

if user_string[0] == '-':
    # check if the rest is all digits
    if user_string[1:].isdigit():
        print("Valid")
    else:
        print("Invalid")
else:
    if user_string.isdigit():
        print("Valid")
    else:
        print("Invalid")
```
